## Setup

Run these cells. They load the libraries, load and clean the data with the same function as the pre-class tutorial, and keep just the four columns you need, in a DataFrame called `bcd2`. You don't need to change anything here.

````python
import numpy as np
import pandas as pd
import seaborn as sns
````

````python
def bcd_load_clean():
    bcd = pd.read_csv('./data/breast_cancer_data.csv')
    bcd['patient_id'] = bcd['patient_id'].astype('string')
    bcd['doctor_name'] = bcd['doctor_name'].str.split().str[1]
    bcd['bare_nuclei'] = bcd['bare_nuclei'].replace('?', '')
    bcd['bare_nuclei'] = pd.to_numeric(bcd['bare_nuclei'])
    
    return bcd
````

````python
bcd = bcd_load_clean()
bcd2 = bcd[['doctor_name', 'clump_thickness', 'bland_chromatin', 'class']].copy()
bcd2.head()
````
