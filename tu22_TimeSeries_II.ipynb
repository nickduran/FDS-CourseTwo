## Summary

When a DataFrame has dates as its index, pandas lets you ask the same data questions at different time scales: `resample` changes the time step (hours to weeks), `rolling` smooths over a window, and grouping by `index.hour` or `index.dayofweek` folds many days into one typical day or week. Each choice shows some patterns and hides others, so the question you are asking decides which one to use.
