## Why Jupyter Notebooks?

Jupyter notebooks are a great option for a class like this for a couple of main reasons. First, the nature of Jupyter notebooks encourage short chunks of code (in code cells) interspersed with explanatory text (markdown cells). Moreover, when run, the code cells can produce output (such as data graphics) directly in the notebook. This makes Jupyter notebooks great for both teaching and learning coding, because both the code and the results can be presented and digested in small bites.

Second, because 

* the markdown text and the code output are interwoven in a single document, and
* this document can be output in a number of formats including pdf and html,

Jupyter Notebooks have become a defacto standard tool for data scientists. For small to intermediate sized projects, everything can be done in a notebook or a small number of notebooks. For larger projects involving lots of complicated code, the projects itself can be done in something like VS Code or PyCharm, and results and summary can still be presented in Jupyter Notebooks simply by reusing the plotting code from the main project.

More minor reasons include the simple easy-to-master interface and ready availability of help due to its ubiquity.

For these reasons, Jupyter notebooks provide us with an ideal jumping in point for learning Python and some fundamentals of data science!

### Telling code cells from markdown cells at a glance

Once a markdown cell has been **rendered** — that is, run — it looks like ordinary document
text, and the difference from a code cell is obvious. The confusing case is a markdown cell
that has *not* been rendered yet. While you are typing in it, it is a plain grey box, and a
grey box is also exactly what an empty code cell looks like.

There is one reliable tell, and it is on the left-hand side:

| What you see on the left | What the cell is |
|---|---|
| `[ ]:` &nbsp;/&nbsp; `[*]:` &nbsp;/&nbsp; `[7]:` | a **code** cell |
| nothing at all | a **markdown** cell |

Only code cells get those brackets, because only code cells can be executed and numbered.
If there is no `[ ]:`, you are looking at markdown, no matter how grey the box is.

And the brackets say more than just "this is code":

| You see | It means |
|---|---|
| `[ ]:` | this cell has **never run** in this session |
| `[*]:` | it is **running right now** |
| `[7]:` | it ran, and it was the **7th** cell run since the kernel started |

Those numbers are the honest record of what your notebook actually did. Reading order is
top to bottom; execution order is whatever you actually clicked. If the numbers do not
count 1, 2, 3 straight down the page, your notebook did not run the way it reads.

## Edit mode and Command mode

This is the single idea that separates people who find Jupyter smooth from people who find
it maddening, and it takes about a minute to learn.

At any moment, a Jupyter notebook is in one of two modes. The difference is simply
**where your keystrokes go**:

| Mode | What your keys do |
|---|---|
| **Edit mode** | Your keys type text **into** the cell. It behaves like any text box anywhere. |
| **Command mode** | Your keys are **single-letter commands about cells** — make one, delete one, change its type. Nothing you type goes into the cell. |

So in edit mode, pressing `b` puts the letter *b* in your code. In command mode, pressing
`b` creates a whole new cell **b**elow. Same key, completely different result, and the only
thing that decides which you get is the mode.

**How to switch:**

| To go to | Press |
|---|---|
| Command mode | `esc` |
| Edit mode | `enter` (or `return`) |
| Edit mode, on a *rendered* markdown cell | `enter`, or just **double-click** the text |

**How to check which mode you are in**, if you are not sure — click on a cell and look at
the border:

- A **green** border with a blinking cursor inside the cell → **edit** mode.
- A **blue** border with no cursor → **command** mode.

The colour is how you *confirm* the mode. It is not what the mode *is*. The mode is about
where your keystrokes land.

> **The mistake absolutely everyone makes once**
>
> You want a new cell, so you press `b`. Instead of a new cell appearing, a stray letter
> `b` lands in the middle of your code, and now that cell throws a `SyntaxError` you did
> not write.
>
> Nothing is broken. You were in **edit** mode, so `b` was just a letter. Delete the stray
> `b`, press `esc` to get to command mode, and press `b` again — this time it makes the
> cell.
>
> Whenever a single-letter shortcut "does nothing" or types a character at you, that is
> the diagnosis: **wrong mode, press `esc` first.**

### The command-mode shortcuts worth knowing

Mastering command mode is what really speeds you up. These are the ones that earn their
keep — every one of them only works **in command mode**, so `esc` first.

| Key | What it does |
|---|---|
| `a` | insert a new cell **a**bove the current one |
| `b` | insert a new cell **b**elow the current one |
| `m` | convert the current cell to **m**arkdown |
| `y` | convert the current cell back to code |
| `d` `d` | **d**elete the current cell (press `d` twice, quickly) |
| `z` | undo the last cell operation — including a delete |
| `c` | **c**opy the current cell |
| `x` | cut the current cell |
| `v` | paste below the current cell |
| `shift`+`v` | paste **above** the current cell |
| `shift`+`enter` | run the cell and move to the next (works in **both** modes) |

Two notes worth having:

- `z` is your undo. If `d` `d` deletes the wrong cell, `z` brings it back. Learn this one
  before you learn `d` `d`.
- Why `y` for "code"? Genuinely nobody knows. It is next to `u` and `t` on the keyboard and
  that is about the best explanation available.

You can see the full list — and search it — by clicking the little keyboard icon in the
toolbar, or under **Settings → Advanced Settings → Keyboard Shortcuts**. Menu items that
have a shortcut show it next to the item.

And, as you have probably worked out already: all the normal keyboard shortcuts for your

### Step 1: Restart and run everything (don't skip this one)

Jupyter lets you run cells in any order, as many times as you like. That flexibility is genuinely useful while you work, but it has a catch: **what is currently on your screen may not match what your code would produce if run cleanly from top to bottom.** You might have edited a cell after running it, deleted something a later cell depends on, or run cells out of order.

So before you submit, always do this:

> **Run → Restart Kernel and Run All Cells…**

This wipes Python's memory and re-runs your whole notebook from the first cell to the last. If everything still works, the outputs saved in your file are trustworthy. If something breaks, you have just found a real problem in your work — and you found it before your instructor did.

### A note on PDFs

Earlier versions of this notebook walked through three different ways to turn a notebook into a PDF. We no longer ask you to do that, and it is worth knowing why.

Every route from a notebook to a PDF is a *conversion*, and each one needs extra software installed on your machine — Pandoc, and for the best-looking output, a full LaTeX distribution (a multi-gigabyte download). On any given laptop one of those pieces is liable to be missing, the wrong version, or impossible to install without administrator rights. Across a class this size, that reliably leaves a handful of people unable to hand in work they have actually completed, through no fault of their own.

Submitting the `.ipynb` file skips all of that. There is no conversion step, so there is nothing to break.

If you ever *do* need a PDF of a notebook — a job application, a poster, an advisor who asks for one — the route that needs no extra software is to print from your browser: **File → Print** in JupyterLab, then set the destination to "Save as PDF". The formatting is imperfect, so check that every cell actually made it into the result before you rely on it.
