## Exercise 5: Who is high-anxiety, in each version?

The lab sorts anxiety scores into three categories with this rule: below 8 is **Low**, from 8 up to (but not including) 12 is **Moderate**, and 12 or above is **High**. The function below does that for one score, using the `if`/`elif`/`else` pattern from the *Ifs, loops, and functions review*. **Run it; you don't need to change it.**

````python
def categorize_anxiety(score):
    if score < 8:
        return "Low"
    elif 8 <= score < 12:
        return "Moderate"
    else:
        return "High"
````

To run that function on **every** participant at once, pandas has a method called `.apply()`. It is new today. It takes a function, runs it on each value in a column, and collects the results into a new column. Here it is for `df_dropped`:

````python
df_dropped["Anxiety_Category"] = df_dropped["Anxiety_Score"].apply(categorize_anxiety)
df_dropped
````
